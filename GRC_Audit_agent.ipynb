```python
# GRC Audit Assistant (Agentic Workflow)

#This project simulates an AI-driven GRC audit assistant using a prompt-chained workflow:

#1. Query Understanding Agent
#2. Retrieval Agent
#3. Analysis Agent

#It maps audit queries to security controls, risks, and compliance frameworks like ISO 27001 and SOC2.
```

```python
knowledge_base = [
    {
        "control_id": "AC-01",
        "title": "Access Control Policy",
        "description": "Users must have unique IDs and role-based access",
        "risk": "Unauthorized access",
        "compliance": ["ISO 27001", "SOC2"]
    },
    {
        "control_id": "EN-01",
        "title": "Data Encryption",
        "description": "Sensitive data must be encrypted at rest and in transit",
        "risk": "Data breach",
        "compliance": ["ISO 27001", "GDPR"]
    },
    {
        "control_id": "LG-01",
        "title": "Logging and Monitoring",
        "description": "System activity must be logged and monitored",
        "risk": "Undetected incidents",
        "compliance": ["SOC2"]
    }
]

print("Knowledge Base Loaded")
```

```python
def extract_keywords(query):
    query = query.lower()
    keywords = []

    if "access" in query:
        keywords.append("access")
    if "data" in query or "encryption" in query:
        keywords.append("data")
    if "log" in query or "monitor" in query:
        keywords.append("log")

    return keywords
```

```python
def retrieve_controls(keywords, kb):
    results = []

    for item in kb:
        text = (item["title"] + item["description"]).lower()
        if any(k in text for k in keywords):
            results.append(item)

    return results
```

```python
def generate_response(query, controls):
    if not controls:
        return {
            "query": query,
            "summary": "No relevant controls found.",
            "controls": [],
            "risks": [],
            "recommendation": "Refine query or expand control database.",
            "compliance": []
        }

    risks = list(set(c["risk"] for c in controls))
    compliance = list(set(sum([c["compliance"] for c in controls], [])))

    return {
        "query": query,
        "summary": "Relevant security controls identified based on query.",
        "controls": [c["control_id"] for c in controls],
        "risks": risks,
        "risk_level": "High" if "breach" in str(risks).lower() else "Medium",
        "recommendation": "Implement and review the above controls regularly.",
        "compliance": compliance
    }
```

```python
from pprint import pprint

query = input("Enter your audit query: ")

# Agent workflow
keywords = extract_keywords(query)
controls = retrieve_controls(keywords, knowledge_base)
response = generate_response(query, controls)

print("\n=== STRUCTURED AUDIT RESPONSE ===\n")

pprint(response)
```

```python
knowledge_base = df.to_dict(orient="records")

knowledge_base
```

```python
def retrieve_controls(keywords, kb):
    results = []

    for item in kb:
        text = (item["Control Name"] + item["Description"]).lower()

        if any(k in text for k in keywords):
            results.append(item)

    return results

```

```python
def generate_response(query, controls):
    if not controls:
        return {"message": "No relevant controls found"}

    return {
        "query": query,
        "controls": [c["Control ID"] for c in controls],
        "evidence_required": [c["Evidence Required"] for c in controls],
        "owners": [c["Owner"] for c in controls],
        "status": [c["Status"] for c in controls]
    }

```

```python
query = "How is data protected?"

keywords = extract_keywords(query)
controls = retrieve_controls(keywords, knowledge_base)
response = generate_response(query, controls)

from pprint import pprint
pprint(response)
```
